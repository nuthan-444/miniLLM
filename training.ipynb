Opening input.txt file for input

```python
with open('input.txt', 'r', encoding='utf-8') as f:
    text = f.read()
```

```python
print("length of dataset in characters: ", len(text))
```

```python
print(text[:100])
```

```python
chars = sorted(list(set(text)))
vocab_size = len(chars)
print("All Unique Characters:", ''.join(chars))
print("Count of Unique Characters:", vocab_size)
```

```python
stoi = { ch:i for i,ch in enumerate(chars) }
iots = { i:ch for i,ch in enumerate(chars) }
encode = lambda str : [ stoi[c] for c in str ]
decode = lambda token : [ iots[str] for str in token ]

print(encode("Hello"))
print(decode(encode("Hello")))
```
